# MarIA

```python
from transformers import AutoTokenizer, AutoModelForTokenClassification

# MarIA
model_name = "PlanTL-GOB-ES/roberta-base-bne"
print(f"Descargando modelo y tokenizer para MarIA...")

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForTokenClassification.from_pretrained(model_name, num_labels=2)  # Ajusta num_labels según tus etiquetas

# Guardar localmente
tokenizer.save_pretrained("./MarIA_tokenizer")
model.save_pretrained("./MarIA_model")

print("MarIA descargado y guardado localmente.")
```
